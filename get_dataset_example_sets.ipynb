```python
from google.colab import files, drive
import numpy as np
import pandas as pd
from sklearn.model_selection import train_test_split

drive.mount('/content/drive')
```

```python
def stratified_subsets(df, label_col):
    remaining = df.copy()
    subsets = []

    for i in range(9):
        subset, remaining = train_test_split(
            remaining,
            train_size=15,
            stratify=remaining[label_col],
            random_state=43 + i
        )
        subset = subset.copy()
        subset['seed_id'] = i + 1
        subsets.append(subset)
    return subsets, remaining
```

```python
files.upload()
qnli_val = pd.read_csv('qnli_val.csv')

existing_ids_qnli = np.load('qnli_few_shot_example_idx.npy').reshape(1, -1)

existing_qnli_subset = qnli_val.loc[existing_ids_qnli.flatten()].copy()
existing_qnli_subset['seed_id'] = 0

df_remaining_qnli = qnli_val[~qnli_val.index.isin(existing_ids_qnli.flatten())]
new_subsets_qnli, clean_remaining_qnli = stratified_subsets(df_remaining_qnli, label_col="label")

all_subsets_qnli = pd.concat([existing_qnli_subset] + new_subsets_qnli, ignore_index=True)

all_subsets_qnli.to_csv("qnli_few_shot_subsets.csv", index=False)
clean_remaining_qnli.to_csv("qnli_val_clean.csv", index=False)
```
